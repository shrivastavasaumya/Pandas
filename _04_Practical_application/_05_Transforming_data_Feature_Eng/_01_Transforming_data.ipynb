```python
import pandas as pd
df = pd.read_csv("raw_data.csv")
```

## Feature Engineering
### To transform, create & add some new columns, new features, that improve the performance of our models


## 1) Apply

```python
#        >50k => 20%                   <--- For 'tax' column income slab
#        <= 50k => 10%

df2 = df.copy()

df2['tax'] = df2['income'].apply(lambda x : "20%" if x >= 50000 else "10%")

df2
```

## 2) Map

```python
# existing values new values k saath map krna chahte hai
```

```python
gender_map = {"Male" : "M", "Female" : "F", "Unknown" : "U"}
df2['gender'] = df2['gender'].map(gender_map)     # to save the values in copy
```

```python
df2
```

## 3) Assign 
### Can create new columns by assigning them some values

```python
# suppose; increase salary by 1.10  

df2 = df2.assign(new_income = df2['income'] * 1.1)

# new column with new value formed
```

## 4) Replace
### replace specific old values with new values

```python
# replace USA by US
df2["country"] = df2["country"].replace('USA','US')     # df2 main isliye save kia taaki yeh original ki jo copy hai, usme main save ho jaay
df2
```

## 5) Rename
#### Can rename the name of rows and columns

```python
# To change head of all columns

df2.columns = ['Id', 'Name', 'Age', 'Country', 'Gender', 'Income', 'Tax', 'New_Income']  # (Just changing the case for example)


# To change head of particular column
df2.rename(columns = {"Income" : "Salary"})


# To change label of particular row
df2.rename(index= {1 : 'First'})
```

## 6) Sort

### sort_values => Based on Values

```python
# sort_values
df2['Income'].sort_values()    # will view sort of only the income column

# To get this change with whole data
df2.sort_values('Income')

# For ascending / descending order
df2.sort_values('Income', ascending= False)

# To sort using two parameters
sorted_df = df2.sort_values(['Income', 'Age'])      # Fact : If two people with same income, then will be ordered according to age
```

### sort_index => Based on index

```python
# sort_index
sorted_df.sort_index()     
```

## 7) Reset 

### Preserves original index, and will create a new column for new indexes after reset

```python
# Preserves original index, and will create a new column for new indexes after reset
sorted_df.reset_index()

# To drop/delete original index, and to have new column for new indexes after reset
sorted_df.reset_index(drop = True)
```

## 8) Rank
### From highest to lowest

```python
sorted_df["Ranking"] = sorted_df['Income'].rank() 
sorted_df        # smallest is getting rank '1'
# clashing/ties/same values will get the average rank of both of them



sorted_df["Ranking"] = sorted_df['Income'].rank(ascending = False) 
sorted_df       # largest is getting rank '1'



sorted_df["Ranking"] = sorted_df['Income'].rank(ascending = False, method = 'dense')  # to avoid average values in ties
sorted_df



sorted_df["Ranking"] = sorted_df['Income'].rank(ascending = False, method = 'min')  # to avoid average values in ties and get minimum value assigned to them
sorted_df



sorted_df["Ranking"] = sorted_df['Income'].rank(ascending = False, method = 'max')  # to avoid average values in ties and get maximum value assigned to them
sorted_df
```

## 9) Reorder
### To reorder columns

```python

# To get 'tax' column at last 
sorted_df[['Id', 'Name', 'Age', 'Country', 'Gender', 'Income', 'New_Income', 'Tax']]

```
