```python
import pandas as pd
```

```python

df = pd.DataFrame({
     "country" : ['USA', 'USA', 'India','India'],
    "year" : [2020, 2021, 2020, 2021],
    'sales' : [100, 120, 90, 110],
    'profit': [20, 25, 18, 22],
    'tax' : [10, 20, 30, 40]
})
```

```python
df
```

```python
melted_df = df.melt(
    id_vars=["country", 'year'],  # want as it is
    value_vars=['sales', 'profit', 'tax'],  # to melt
    var_name='metrics',  # name for new column
    value_name='value'    # name for the new column having actual data values
)

melted_df
```

```python
original = melted_df.pivot(
    index= ['country', 'year'],   # we use a pair or a name which is unique for every single element to separate it in single rows
    columns = 'metrics',          # apne andr hr type k head/label ka separate naya column banayga  
    values = 'value'              # inn values se upr k columns ko fill krna chahte hai
)
```
