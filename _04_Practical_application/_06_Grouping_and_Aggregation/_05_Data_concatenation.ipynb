## Data Concatenation

### Stacking dataframes on top of each other (rows) or side by side (columns)

```python
# data ko saath-saath chipka de, ya fir ek k niche ek jod de
```

```python
df1 = pd.DataFrame({
    'id' : [1, 2, 3],
    'name' : ['Adam', 'Biman', 'Charlie']
})

df2 = pd.DataFrame({
    'id' : [ 4, 5, 6],
    'name' : ['Dora', "Nora", "Lora"]
})
```

```python
# STACKING DATA ON TOP OF EACH OTHER

pd.concat([df1, df2])     # row-wise concat, here the old index is preserved


pd.concat([df1, df2], ignore_index = True)     # if we want to change old index/label into new ones

```

```python
# will be stacked in the form of wide table

pd.concat([df1, df2], axis = 1)    # column wise concat
                                   # jo 2nd table hai vo add ho jaygi 1st main as a form of new columns
```
