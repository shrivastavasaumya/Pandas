```python
import pandas as pd
df = pd.read_csv("raw_data.csv")
```

## groupby()
### a powerful tool used to split data into groups based on specific criteria, apply operations to those groups, and then combine the results

```python
df
```

```python

# hr country ki alg group hoga[group] + hr group k income ka mean[aggregation]
df.groupby("country")['income'].mean()   # mean income of each group 
df.groupby("country")['income'].min()    # min income of each group 
df.groupby("country")['income'].max()    # max income of each group 
```

```python
# konsa aisa parameter hai jiske basis pr hum group bana rahe hai = groupby() {here parameter => 'gender'}

df.groupby("gender")['income'].mean() 

```

## agg()
### used to apply one or more operations (aggregations) across a specified axis of a DataFrame or Series

```python
# groups pr multiple function apply krne k liye hota hia agg()

#Example-1

df.groupby("gender")['income'].agg(['mean', 'min', 'max']) 
```

```python
# Exapmle-2

df.groupby("country")['income'].agg(['mean', 'min', 'max']) 

# aggregate(): an alias (an alternative name for an existing object, module, or type) for agg()

df.groupby("country")['income'].aggregate(['mean', 'min', 'max']) 
```

## Renaming + groupby() + agg()

```python
# TO CHANGE HEAD / LABELS TO RENAME THESE COLUMNS  [mean , min , max]

df.groupby("country")['income'].aggregate(avg_salary = 'mean', min_salary = 'min', max_salary = 'max') 
```

```python
# alg-alg columns pr alg-alg aggregate functions ko apply kr skte hai

# Example-1
df.groupby("country").agg({
   "income" : "mean",
    "age" : "mean"    
}) 



# Example-2
df.groupby("country").agg({
   "income" : "max",
    "age" : "mean"    
})

```

```python
# renaming columns

# Example-3
df.groupby("country").agg(    # to add labels/ rename columns
   max_salary = ("income" , "max"),
   avg_age = ("age" , "mean" )  
)
```
