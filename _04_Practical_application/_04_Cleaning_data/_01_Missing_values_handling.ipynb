### dropna()

```python
# jis bhi row main koi bhi value 'NaN' hui toh yeh poori row drop(eliminate) kr deta hai

df.dropna()
```

```python
df.dropna(axis = 1)     # jis bhi column main koi bhi value 'NaN' hui toh yeh poori column drop(eliminate) kr deta hai
```

### fillna(value)

```python
# filling ( blank/NaN ) none values with given value

df.fillna(100)
```

```python
# To to fill the missing values under 'age' with the mean value of age

age_mean = df["age"].mean()
df['age'].fillna(age_mean)
```

```python

# want to save as a 'copy' and then, process it  = apne cleaned data k liye original data ki copy banayenge

cleaned_data = df.copy()
age_mean = df["age"].mean()
cleaned_data['age']= cleaned_data['age'].fillna(age_mean)

cleaned_data
```

```python
df    # original values are unchanged
```

### ffill()  == Forward fill

```python
# upar ki value ko exact niche ki misising value main fill kr deta hai

df.ffill()
```

### bfill() == Backward fill

```python
# niche ki values se exact upar ki missing values ko fill kr deta hai

# niche se upar

df.bfill()
```
