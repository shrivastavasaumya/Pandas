```python
import pandas as pd
df = pd.read_csv("raw_data.csv")
```

#### Lower Casing

```python
# Take a column with string 

df['gender'].str.lower()
```

#### Upper Casing

```python
df['name'].str.upper()
```

#### Capitalize

```python
df['name'].str.capitalize()         # Capitalize the start of each word
```

## Split 
### allows you to split strings using almost any character, sequence of characters, or regular expression

```python
df['name'].str.split(" ")  #---> we used an empty string as an separator

# Here; Jis jagah pr space hai vaha se yeh unn words ko tukdo main baat dega [separating 1st and last name]
```

### Contains

```python
# To check whether the value exists or not (contains)

df['country'].str.contains('US')
```

```python
# To check irrespective of case


df['country'].str.contains('india', case = False)
```
