```python
import pandas as pd

# Creating a DataFrame
data = {
    "Name": ["John", "Alice", "Bob"],
    "Age": [25, 30, 22],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

print(df)
```

```python
df   # for proper visualization
```

```python
print(df.index)       # row indexing
print(df.columns)     # column indexing
```

```python
# create list of lists
dsst = pd.DataFrame([["Aka", 23], ["baka", 24], ["kaka", 26]], columns = ["Name", "Age"])
print(dsst)       # the sub lists are individual rows 
```

# via numpy array

```python
import numpy as np
np_arr = np.array([[1,2,3], [4,5,6]])
df = pd.DataFrame(np_arr, columns = ["A" , "B" , "C"])
print(df)
                  
```
