```python
# Copied and converted 'csv' file to a new 'json' file


import pandas as pd
# read csv file
df = pd.read_csv("employee_data.csv")

# convert to json
df.to_json("employee_data.json", orient="records", indent=4)

print("CSV converted to JSON successfully!")
```

```python
#  How to read a excel file

import pandas as pd
df = pd.read_excel("employee_data.xlsx")

print(df, type(df))
```
