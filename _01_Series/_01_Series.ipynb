## Difference betweeen "LABEL" and "INDEX" in Pandas

```python
#  We can access a value via two-ways

#                                                        1. Index in Pandas
# An index is the left-side identifier of rows in a DataFrame or Series. [position]

data = {
    "Name": ["A", "B", "C"],
    "Marks": [90, 85, 88]
}

df = pd.DataFrame(data)

print(df)
```

```python
#                                                        2. Label in Pandas
# A label is the actual name used to identify rows or columns. [Named index]

# Labels can be:
#          row names
#          column names

df.index = ["adam", "bob", "charlie"]

print(df)
```

```python
line = pd.Series([23, 24, 25, 26], index = ["aka", "baka", "chaka", "naka"])

print(line)
print(line["aka"])
print(line["baka"])  # via label

print(line[0])     # via index
```
