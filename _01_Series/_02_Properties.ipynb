## Properties of Series in Pandas

1) Homogenous

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40, "Laddu"])

print(s)
print(s.dtype)
```

2) Vectorized Operations

```python
s = pd.Series([1, 2, 3, 4])

print(s + 5)
```

3) Handle Missing Values with NaN

```python
# "None" --> "NaN"

s = pd.Series([10, 20, None, 40])

print(s)
```

4) Mutable values, immutable size

```python
s = pd.Series([10, 20, 30])

s[1] = 200
changed_s = s.drop(0)   # Jab bhi hum koi element 'add' ya 'remove' krenge, hume new series ko "changed" series create krni padegi

print(s)
print(changed_s)
```
